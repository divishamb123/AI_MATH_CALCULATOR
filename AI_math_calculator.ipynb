```python
import numpy as np
def input_matrix(name):
    print(f"\nEnter  values for {name} ")
    try:
        return np.array(eval(input()))
    except Exception as e:
        print("invalid")
        return input_matrix(name)

def input_vector(name):
    print(f"\nEnter  values for {name} ")
    try:
        return np.array(eval(input()))
    except Exception as e:
        print("invalid input")
        return input_vector(name)

def matrix_multiply(A,B):
  try:
    return np.dot(A,B)
  except Exception as e:
    return f" error :{e}"

def vector_dot(v1,v2):
  try:
    return np.dot(v1,v2)
  except Exception as e:
    return f" error :{e}"

def sigmoid (x):
  return 1/(1+np.exp(-x))

def relu(x):
  return np.maximum(0,x)

def softmax(x):
  e_x=np.exp(x-np.max(x))
  return e_x/e_x.sum()

def run_calculator():
   print("AI math calculator")

while True:
  print("select an operation")
  print("1.matrix multiplication")
  print("2.Vector Dot Product")
  print("3.sigmoid function")
  print("4.relu function")
  print("5.softmax function")
  print("6.exit")

  choice=input("enter your choices[1-6]")
  if choice == '1':
    A=input_matrix(" matrix A")
    B=input_matrix("matrix B")
    result=matrix_multiply(A,B)
    print("Result:\n",result)

  elif choice == '2':
   v1=input_vector("vector1")
   v2=input_vector("vector2")
   result=vector_dot(v1,v2)
   print("Dot product:\n",result)

  elif choice == '3':
   v=input_vector("input vector")
   result=sigmoid(v)
   print("sigmoid output:\n",result)

  elif choice == '4':
   v=input_vector("input vector")
   result=relu(v)
   print("Relu output:\n",result)

  elif choice == '5':
   v=input_vector("input vector")
   result=softmax(v)
   print("softmax output:\n",result)

  elif choice == '6':
   print("exit")
   break
  else:
   print("invalid choice")

   run_calculator()
```
